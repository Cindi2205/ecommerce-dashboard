## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Metode pembayaran apa yang mendominasi transaksi pada 5 kategori produk dengan total penjualan (revenue) tertinggi di wilayah Sao Paulo selama semester pertama tahun 2018?
- Specific: Fokus pada hubungan antara "5 kategori produk dengan penjualan tertinggi" dan "metode pembayaran" yang digunakan oleh pelanggan di wilayah spesifik "Sao Paulo".
- Measurable: Menggunakan batasan terukur berupa peringkat atau kuantitas ("5 kategori produk" dengan total penjualan tertinggi).
- Action-Oriented: Membantu tim manajemen atau pemasaran untuk mengoptimalkan promo, diskon, atau kerja sama eksklusif (cashback/cicilan) dengan penyedia metode pembayaran tertentu pada kategori produk paling laris.
- Relevant: Pemahaman mengenai preferensi pembayaran pada produk unggulan sangat krusial untuk meningkatkan kenyamanan
checkout dan mengurangi tingkat keranjang belanja yang ditinggalkan (cart abandonment).
- Time-bound: Dibatasi pada periode waktu yang jelas, yaitu "semester pertama tahun 2018" (Januari–Juni 2018).
- **Pertanyaan 2:** Berapa rata-rata selisih waktu keterlambatan pengiriman oleh seller di luar wilayah Rio de Janeiro yang memicu skor ulasan buruk (review score 1) sebesar lebih dari 20% pada kuartal ketiga tahun 2017?
- Specific: Fokus pada "selisih waktu keterlambatan pengiriman" dari "seller di luar wilayah Rio de Janeiro" yang berkorelasi langsung dengan "skor ulasan buruk (review score 1)".
- Measurable: Menggunakan metrik kuantitatif yang jelas, yaitu proporsi ulasan buruk sebesar "lebih dari 20%".
- Action-Oriented: Memungkinkan tim operasional seller management untuk memberikan peringatan, penalti, atau evaluasi logistik kepada seller di luar wilayah tersebut guna memperbaiki ketepatan waktu pengiriman.
- Relevant: Kepuasan pelanggan sangat bergantung pada kecepatan pengiriman; penanganan masalah ini secara langsung melindungi reputasi platform dan meningkatkan retensi pelanggan.
- Time-bound: Dibatasi pada rentang waktu spesifik, yaitu "kuartal ketiga tahun 2017" (Juli–September 2017).

## Exploratory Data Analysis (EDA)

### Explore ...

```python
display(customers_df.sample(5))

display(customers_df.describe(include="all"))

print(f"customer_id unique: {customers_df.customer_id.is_unique}")
print(f"Total baris: {len(customers_df)}, Unique ID: {customers_df.customer_unique_id.nunique()}")

state_summary = customers_df.groupby(by="customer_state").agg({
    "customer_id": ["count", "nunique"]
})
display(state_summary)

display(customers_df.groupby(by="customer_city").customer_id.nunique().sort_values(ascending=False).head(5))

display(customers_df.groupby(by="customer_state").customer_id.nunique().sort_values(ascending=False))

orders_df['order_purchase_timestamp'] = pd.to_datetime(orders_df['order_purchase_timestamp'])
start_date = '2018-01-01'
end_date = '2018-06-30'

df_q1 = orders_df[
    (orders_df['order_purchase_timestamp'] >= start_date) &
    (orders_df['order_purchase_timestamp'] <= end_date)
].merge(customers_df, on='customer_id', how='inner') \
 .merge(order_items_df, on='order_id', how='inner') \
 .merge(products_df, on='product_id', how='inner') \
 .merge(order_payments_df, on='order_id', how='inner')

df_q1_sp = df_q1[df_q1['customer_city'].str.lower() == 'sao paulo']

top_5_categories = df_q1_sp.groupby('product_category_name')['price'].sum().reset_index() \
    .sort_values(by='price', ascending=False).head(5)
display(top_5_categories)

top_categories_list = top_5_categories['product_category_name'].tolist()
df_top5_sp = df_q1_sp[df_q1_sp['product_category_name'].isin(top_categories_list)]

payment_dominance = df_top5_sp.groupby(['product_category_name', 'payment_type'])['order_id'].count().reset_index(name='transaction_count')
display(payment_dominance)

orders_df['order_delivered_customer_date'] = pd.to_datetime(orders_df['order_delivered_customer_date'])
orders_df['order_estimated_delivery_date'] = pd.to_datetime(orders_df['order_estimated_delivery_date'])

q3_start = '2017-07-01'
q3_end = '2017-09-30'

df_q3_2017 = orders_df[
    (orders_df['order_purchase_timestamp'] >= q3_start) &
    (orders_df['order_purchase_timestamp'] <= q3_end)
]

df_q2_merged = df_q3_2017.merge(order_items_df, on='order_id', how='inner') \
    .merge(sellers_df, on='seller_id', how='inner') \
    .merge(order_reviews_df, on='order_id', how='inner')

df_outside_rj = df_q2_merged[df_q2_merged['seller_state'].str.upper() != 'RJ']

# Perbaikan menggunakan .loc agar warning hilang
df_outside_rj.loc[:, 'delivery_delay_days'] = (df_outside_rj['order_delivered_customer_date'] - df_outside_rj['order_estimated_delivery_date']).dt.days

bad_reviews_delayed = df_outside_rj[(df_outside_rj['review_score'] == 1) & (df_outside_rj['delivery_delay_days'] > 0)]

total_reviews_outside = len(df_outside_rj)
total_score1_delayed = len(bad_reviews_delayed)
percentage_score1 = (total_score1_delayed / total_reviews_outside) * 100 if total_reviews_outside > 0 else 0

print(f"Persentase Ulasan Buruk (Score 1) karena Keterlambatan: {percentage_score1:.2f}%")

mean_delay_days = bad_reviews_delayed['delivery_delay_days'].mean()
print(f"Rata-rata selisih waktu keterlambatan pengiriman: {mean_delay_days:.2f} hari")
```

## Visualization & Explanatory Analysis

### Pertanyaan 1:

```python
plt.figure(figsize=(12, 6))

ax1 = sns.barplot(
    data=payment_dominance,
    x='product_category_name',
    y='transaction_count',
    hue='payment_type',
    palette='viridis'
)

plt.title('Dominasi Metode Pembayaran pada Top 5 Kategori Produk di Sao Paulo (H1 2018)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Kategori Produk', fontsize=12)
plt.ylabel('Jumlah Transaksi', fontsize=12)
plt.xticks(rotation=25, ha='right', fontsize=10)
plt.legend(title='Metode Pembayaran', title_fontsize='11', fontsize=10)

for p in ax1.patches:
    height = p.get_height()
    if height > 0:
        ax1.annotate(f'{int(height)}',
                    (p.get_x() + p.get_width() / 2., height),
                    ha='center', va='bottom',
                    fontsize=9, color='black',
                    xytext=(0, 3),
                    textcoords='offset points')

plt.tight_layout()
plt.show()
```

### Pertanyaan 2:

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=bad_reviews_delayed,
    x='delivery_delay_days',
    bins=30,
    kde=True,
    color='crimson'
)

plt.title('Distribusi Waktu Keterlambatan Pengiriman dengan Ulasan Buruk (Score 1) di Luar RJ (Q3 2017)', fontsize=13, fontweight='bold', pad=15)
plt.xlabel('Jumlah Hari Keterlambatan', fontsize=12)
plt.ylabel('Frekuensi Pesanan', fontsize=12)

plt.axvline(
    mean_delay_days,
    color='blue',
    linestyle='--',
    linewidth=2,
    label=f'Rata-rata Keterlambatan: {mean_delay_days:.2f} hari'
)

plt.legend(fontsize=11)
plt.tight_layout()
plt.show()
```

## Conclusion & Recommendation

**Conclusion:**
- Pertanyaan 1: Metode pembayaran yang sangat mendominasi transaksi pada top 5 kategori produk di wilayah Sao Paulo selama semester pertama tahun 2018 adalah credit_card (kartu kredit), dengan jumlah transaksi tertinggi berada pada kategori beleza_saude (704 transaksi) dan cama_mesa_banho (696 transaksi). Posisi berikutnya secara konsisten diikuti oleh metode boleto, voucher, dan debit_card di seluruh kategori produk utama tersebut.
- Pertanyaan 2: Rata-rata selisih waktu keterlambatan pengiriman oleh seller di luar wilayah Rio de Janeiro yang memicu skor ulasan buruk (score 1) pada kuartal ketiga tahun 2017 adalah 10.41 hari. Sebagian besar pesanan dengan ulasan buruk mengalami keterlambatan berkisar antara 1 hingga 10 hari dari estimasi waktu yang dijanjikan.

**Recommendation:**
- Strategi Pembayaran & Promosi: Mengingat dominasi pembayaran menggunakan kartu kredit pada produk terlaris, tim pemasaran dapat memperkuat kerja sama eksklusif dengan penyedia layanan perbankan (seperti promo cicilan 0% atau cashback) guna menjaga kenyamanan checkout dan menekan angka cart abandonment.
- Evaluasi & Mitigasi Logistik Seller: Tim operasional perlu memberikan perhatian khusus atau evaluasi performa logistik kepada seller di luar wilayah Rio de Janeiro. Mengingat keterlambatan rata-rata mencapai lebih dari 10 hari dan memicu ulasan buruk, penerapan sanksi performa atau perbaikan kerja sama dengan mitra ekspedisi di luar wilayah tersebut sangat krusial untuk melindungi reputasi platform.
